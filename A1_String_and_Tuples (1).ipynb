```python
# STRING METHODS.

"""Write a Python program that takes a string, strM = “Python beginner tutorial” and
perform the following tasks:"""
```

```python
strM="Python beginner tutorial"
print(strM)
```

```python
#convert the sentence to uppercase ?
print(strM.upper())
```

```python
#convert the sentence to lowercase ?
print(strM.lower())
```

```python
#capitalize and return the sentence to original input form ?
print(strM.capitalize())
```

```python
#count the total no.of occurrences of "t" in the string ?
print(strM.count("t"))
```

```python
#replace "python" with "Machine Learning"
print(strM.replace("Python","Machine Learning"))
```

```python
#TUPLES (Creating,Modification and Access).
```

```python
#create tuples

tuple1=(10,20,30)
print(tuple1)

tuple2=(40,50,60)
print(tuple2)
```

```python
#concatenate the two tuple & store it in "t_combine" ?
t_combine=tuple1+tuple2
print(t_combine)
```

```python
#Repeat the elements of "t_combine" 3 times?
print(t_combine * 3)
```

```python
#access the 3rd element from "t_combine"?
print(t_combine[2])
```

```python
#access the first three elements from "t_combine"?
print(t_combine[0:3])
```

```python
#access the last three elements from "t_combine"?
print(t_combine[-3:])
```
